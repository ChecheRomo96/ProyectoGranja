# Organización de Datos Crudos

Este notebook automatiza la organización de archivos sin procesar en el directorio `data/raw` clasificándolos por tipo de archivo (CSV, PDF, XLS).

## Objetivo
- Explorar los archivos presentes en el directorio de datos crudos
- Identificar los tipos de archivo disponibles
- Organizar los archivos en subdirectorios según su extensión

## Paso 1: Exploración del Directorio de Datos Crudos

Este paso inicial explora el directorio `data/raw` para identificar qué archivos están presentes y contar los tipos de archivo disponibles.

```python
from pathlib import Path

# Define la ruta del directorio de datos crudos
DATA_DIR = Path("../data/raw")

# Obtiene la lista de todos los archivos en el directorio de manera recursiva
files = list(DATA_DIR.rglob("*"))

print("Total archivos:", len(files))

# Contabiliza los archivos por extensión
extensions = {}

for f in files:
    if f.is_file():  # Asegura que solo sean archivos
        ext = f.suffix.lower()
        extensions[ext] = extensions.get(ext, 0) + 1

# Muestra el resumen de extensiones encontradas
extensions
```

## Paso 2: Listar Archivos CSV

Identifica y enumera todos los archivos con extensión `.csv` en el directorio de datos crudos.

```python
# Busca todos los archivos CSV en el directorio de datos crudos de manera recursiva
csv_files = list(DATA_DIR.glob("**/*.csv"))

print("CSV encontrados:", len(csv_files))
print()

# Enumera el nombre de cada archivo CSV
for f in csv_files:
    print(f.name)
```

## Paso 3: Listar Archivos PDF

Identifica y enumera todos los archivos con extensión `.pdf` en el directorio de datos crudos.

```python
# Busca todos los archivos PDF en el directorio de datos crudos de manera recursiva
pdf_files = list(DATA_DIR.glob("**/*.pdf"))

print("PDF encontrados:", len(pdf_files))
print()

# Enumera el nombre de cada archivo PDF
for f in pdf_files:
    print(f.name)
```

## Paso 4: Listar Archivos XLS

Identifica y enumera todos los archivos con extensión `.xls` y `.xlsx` en el directorio de datos crudos.

```python
# Busca todos los archivos PDF en el directorio de datos crudos de manera recursiva
pdf_files = list(DATA_DIR.glob("**/*.xls"))

print("XLS encontrados:", len(pdf_files))
print()

# Enumera el nombre de cada archivo XLS
for f in pdf_files:
    print(f.name)
```

```python
# Busca todos los archivos XLS en el directorio de datos crudos de manera recursiva
xls_files = list(DATA_DIR.glob("**/*.xls*"))  # Incluye .xls y .xlsx

print("XLS encontrados:", len(xls_files))
print()

# Enumera el nombre de cada archivo XLS
for f in xls_files:
    print(f.name)
```

## Paso 5: Organizar Archivos PDF

Crea una subcarpeta `docs` y mueve todos los archivos con extensión `.pdf` al directorio `data/raw/docs/` para organizarlos de manera separada.

```python
import shutil

# Define la ruta del subdirectorio para documentos
DOCS_DIR = DATA_DIR / "eventos"

# Crea el directorio si no existe
DOCS_DIR.mkdir(exist_ok=True)

# Mueve todos los archivos PDF al subdirectorio de manera recursiva
for pdf in DATA_DIR.glob("**/*.pdf"):
    shutil.move(pdf, DOCS_DIR / pdf.name)
    print(f"Movido: {pdf.name} -> docs/")
```

## Paso 6: Organizar Archivos XLS

Crea una subcarpeta `xls` y mueve todos los archivos con extensión `.xls` y `.xlsx` al directorio `data/raw/xls/` para organizarlos de manera separada.

```python
import shutil

# Define la ruta del subdirectorio para archivos Excel
XLS_DIR = DATA_DIR / "produccion"

# Crea el directorio si no existe
XLS_DIR.mkdir(exist_ok=True)

# Mueve todos los archivos XLS al subdirectorio de manera recursiva
for xls in DATA_DIR.glob("**/*.xls*"):  # Incluye .xls y .xlsx
    shutil.move(xls, XLS_DIR / xls.name)
    print(f"Movido: {xls.name} -> xls/")

# Scar el excel de la alimentacion y moverlo a su carpeta correspondiente
shutil.move("../data/raw/produccion/sobrantes y dietas 2025.xlsx", "../data/raw/dieta/sobrantes y dietas 2025.xlsx")
```

## Paso 7: Organizar Archivos CSV

Crea una subcarpeta `csv` y mueve todos los archivos con extensión `.csv` al directorio `data/raw/csv/` para organizarlos de manera separada.

```python
import shutil

# Define la ruta del subdirectorio para archivos CSV
CSV_DIR = DATA_DIR / "rumia"

# Crea el directorio si no existe
CSV_DIR.mkdir(exist_ok=True)

# Mueve todos los archivos CSV al subdirectorio de manera recursiva
for csv in DATA_DIR.glob("**/*.csv"):
    shutil.move(csv, CSV_DIR / csv.name)
    print(f"Movido: {csv.name} -> csv/")
```

## Conclusión

El notebook ha completado la organización de los archivos crudos. Los archivos ahora están organizados en las siguientes subdirectorios:

- `data/raw/csv/` - Contiene todos los archivos CSV (datos estructurados)
- `data/raw/docs/` - Contiene todos los documentos PDF (referencias y documentación)
- `data/raw/xls/` - Contiene todos los archivos Excel (.xls y .xlsx) (hojas de cálculo)

Esta estructura facilita la navegación y el acceso a los diferentes tipos de datos durante el análisis del proyecto.
